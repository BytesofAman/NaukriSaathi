```python
import re
import nltk 
from nltk.stem import WordNetLemmatizer
from nltk import word_tokenize
```

```python
nltk.download("punkt")
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
job_domains = {
    "data analyst": [
        "python", "sql", "excel", "pandas", "power bi"
    ],
    "machine learning engineer": [
        "python", "machine learning", "numpy", "pandas", "tensorflow"
    ],
    "web developer": [
        "html", "css", "javascript", "react"
    ]
}
```

#### 3) Text Cleaning & Lemmatization

```python
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    text = text.lower()
    text =re.sub(r'[^a-zA-Z]', ' ' ,text)
    tokens = word_tokenize(text)
    lemmas = [lemmatizer.lemmatize(token) for token in tokens]
    return lemmas
```

```python
def lemmatize_skills(skill_list):
    lemmatized = []

    for skill in skill_list:
        tokens = skill.lower().split()
        lemmas = [lemmatizer.lemmatize(token) for token in tokens]
        lemmatized.append(" ".join(lemmas))

    return lemmatized
```

```python
def extract_skill(user_text , skill_list):
    extracted = set()
    user_tokens = preprocess_text(user_text)
    # user_text_clean = " ".join(user_tokens)
    lemmatized_skills = lemmatize_skills(skill_list)
    
    for original_skill , lemma_skill in zip(skill_list , lemmatized_skills):
        skill_tokens = lemma_skill.split()
        if all(token in user_tokens for token in skill_tokens ):
                extracted.add(original_skill)

    return list(extracted)
```

#### 4) Job-Fit Matching

```python
def check_job_filter(user_skills,required_skills):
    matched = set(user_skills) & set(required_skills)
    missing = set(required_skills) - set(user_skills)
    score = (len(matched) / len(required_skills)) * 100
    return score , matched , missing
```

#### 5) Chatbot Response Rules

```python
def chatbot_reponse(score):
    if score >= 70:
        return "You are a Good fitted For this job role."
    elif score >= 40:
        return "You are partially fit for this job role."
    else:
        return "You are not fit for this job role yet "

```

```python
DOMAIN_HOBBIES = {
    "data analyst" : ["chess" , "reading" , "coding" , "puzzles"],
    "ml engineer"  : [" coding" , "research" , "math"],
    "web developer"  : [" coding" , "design" , "blogging"]
}

def hobby_score(hobbies_text , domain):
    tokens = [h.strip().lower() for h in hobbies_text.split(",")]
    domain_hobbies = DOMAIN_HOBBIES.get(domain ,[])
    matched = set(tokens) & set(domain_hobbies)
    return len(matched) * 2 # weight per match
```

```python
SOFT_SKILL_KEYWORDS = ["leadership" , "teamwork" , "communication" , "problem solving" , "adaptability"]

def soft_skill_score(text):
    text = text.lower()
    score = 0
    for skill in SOFT_SKILL_KEYWORDS:
        if skill in text:
            score += 2

    return score
```

```python
def academic_score(tenth , twelfth):
    return (tenth * 0.4 + twelfth * 0.6)
```

```python
print("Hello! I am your NaukriSaathi , A ChatBot\n")

print("Available Domains are:")
for domain in job_domains:
    print("-" ,domain)

domain = input("\n Enter domain you want to apply for:")
domain = re.sub(r'\s+', ' ', domain).strip().lower()

# user_skills = [
#     skill.strip().lower()
#     for skill in domain.split(",")
# ]


if domain not in job_domains:
    print("Invalid Domain Please Select Valid Domains Only. ")
    exit()

else:
    user_text = input("\n Describe your skills (example: I know Python , SQL ,and Pandas):\n ")

required_skills = job_domains[domain]
user_skills = extract_skill(user_text , required_skills)
score, matched, missing = check_job_filter(user_skills, required_skills)


print("\n----------- Skill Analysis Result --------")
print(chatbot_reponse(score))
print(f"Match Perccentage: {score:.2f}%")

print("\n Matched Skills:")
for skill in matched:
    print("-", skill)

print("\n Missing Skills:")
for skill in missing:
    print("-",skill)
    
print("\n ---- Next Step : Academic and Personal Deatails ----\n")

tenth = float(input("Enter your 10th percentage: "))
twelfth = float(input("Enter your 12th percentage: "))

hobbies = input("Enter your hobbies (comma separated)")

strengths = input("Enter your Strenght: ")
weakness = input("Enter your Weakness: ")

academic = academic_score(tenth , twelfth)
hobby = hobby_score(hobbies , domain)
strength_score = soft_skill_score(strengths)
weakness_score = soft_skill_score(weakness)

final_score = (
    score* 0.5 +
    academic * 0.3 + 
    (hobby + strength_score - weakness_score)*0.2
)

print("\n--- Final Candidate Recommendation ---")
print(f"Technical Score: {score:.2f}%")
print(f"Academic Score: {academic:.2f}")
print(f"Hobby + Soft Skills Score: {hobby + strength_score - weakness_score:.2f}")
print(f"Final Composite Score: {final_score:.2f}")

if final_score >= 70:
    print("Recommendation: Strong Fit")
elif final_score >= 40:
    print("Recommendation: Partial Fit")
else:
    print("Recommendation: Not Fit")
```
